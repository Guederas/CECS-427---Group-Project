### Variable Guide (quick reference)

General:
- `G`: directed graph (`nx.DiGraph`)
- `nodes`: list of node labels

HITS:
- `auth[p]`: authority score of page p
- `hub[p]`: hub score of page p
- **Authority update:** new_auth[p] = sum of hub scores of nodes that link **to** p
- **Hub update:** new_hub[p] = sum of authority scores of nodes that p links **to**
- Normalization in this assignment: divide each vector by its **sum** so it sums to 1.

PageRank:
- `pr[p]`: PageRank value of page p
- `n`: number of nodes
- `outdeg(p)`: number of outgoing links
- **Basic update:** each node splits its PageRank equally among outgoing links; a node with outdeg=0 keeps all its PageRank (treat as linking to itself).
- `s`: scaling factor for scaled PageRank (0 < s < 1); after basic update, multiply by `s` and add `(1-s)/n` to every node.

## Part C — Leakage and Scaled PageRank

Basic PageRank can “leak” into dead-end regions that can be reached but can’t link back.

Scaled PageRank fixes this by:
1. Apply the basic update.
2. Scale everything by `s` (0 < s < 1).
3. Redistribute the remaining `(1-s)` equally: each node gets `(1-s)/n`.
